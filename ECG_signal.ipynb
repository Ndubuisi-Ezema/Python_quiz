```python
## Detect R waves using find_peaks

from scipy.signal import find_peaks

# Extract the first 3 seconds of Lead II
lead_ii = df["II"].values[:3 * fs]

# Detect R peaks
# height: minimum peak amplitude
# distance: minimum separation between peaks (0.5 s = 250 samples)
peaks, properties = find_peaks(
    lead_ii,
    height=500,      # Adjust if necessary
    distance=250     # Prevents multiple detections within one heartbeat
)

# Print number of detected R peaks
print("Total number of detected R peaks:", len(peaks))

# Plot ECG and detected R peaks
time = [i / fs for i in range(len(lead_ii))]

plt.figure(figsize=(12, 4))
plt.plot(time, lead_ii, label="Lead II ECG")
plt.plot(time[0:len(lead_ii)], lead_ii)

# Highlight R peaks
plt.scatter(peaks / fs, lead_ii[peaks],
            color='red', marker='o', s=60, label='R Peaks')

plt.title("Lead II ECG (First 3 Seconds) with Detected R Peaks")
plt.xlabel("Time (s)")
plt.ylabel("Voltage (µV)")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
## Heart rate estimation

# Sampling frequency (Hz)
fs = 500

# Extract first 3 seconds of Lead II
lead_ii = df["II"].values[:3 * fs]

# Detect R peaks
peaks, properties = find_peaks(
    lead_ii,
    height=500,      # Adjust if necessary
    distance=250     # Minimum distance between R peaks
)

# -------------------------------
# (a) Calculate RR intervals
# -------------------------------
# RR intervals in samples
rr_intervals_samples = np.diff(peaks)

# Convert RR intervals to seconds
rr_intervals_seconds = rr_intervals_samples / fs

print("RR Intervals (seconds):")
print(rr_intervals_seconds)

# -------------------------------
# (b) Average RR interval
# -------------------------------
average_rr = np.mean(rr_intervals_seconds)

print("\nAverage RR Interval: {:.3f} seconds".format(average_rr))

# -------------------------------
# (c) Estimate Heart Rate (BPM)
# -------------------------------
heart_rate = 60 / average_rr

print("Estimated Heart Rate: {:.2f} BPM".format(heart_rate))
```

```python
### Apply median filter

from scipy.signal import medfilt

# Extract first 3 seconds of Lead II
duration = 3  # seconds
num_samples = fs * duration

lead_ii = df["II"].values[:num_samples]

# Time axis
time = np.arange(num_samples) / fs

# ----------------------------------------
# Apply Median Filters
# ----------------------------------------
filtered_k3 = medfilt(lead_ii, kernel_size=3)
filtered_k11 = medfilt(lead_ii, kernel_size=11)

# ----------------------------------------
# Plot Results
# ----------------------------------------
plt.figure(figsize=(12, 10))

# Original ECG
plt.subplot(3, 1, 1)
plt.plot(time, lead_ii, color='blue', linewidth=1)
plt.title("Original Lead II ECG (First 3 Seconds)")
plt.ylabel("Voltage (µV)")
plt.grid(True)

# Median Filter (Kernel = 3)
plt.subplot(3, 1, 2)
plt.plot(time, filtered_k3, color='green', linewidth=1.2)
plt.title("Median Filtered ECG (Kernel Size = 3)")
plt.ylabel("Voltage (µV)")
plt.grid(True)

# Median Filter (Kernel = 11)
plt.subplot(3, 1, 3)
plt.plot(time, filtered_k11, color='red', linewidth=1.2)
plt.title("Median Filtered ECG (Kernel Size = 11)")
plt.xlabel("Time (s)")
plt.ylabel("Voltage (µV)")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
### Apply Gaussian filter

from scipy.ndimage import gaussian_filter1d

# Extract first 3 seconds of Lead II
duration = 3  # seconds
num_samples = fs * duration

lead_ii = df["II"].values[:num_samples]

# Create time axis
time = np.arange(num_samples) / fs

# Apply Gaussian Filters with different sigma values
sigma1 = 1
sigma2 = 3
sigma3 = 5

filtered_sigma1 = gaussian_filter1d(lead_ii, sigma=sigma1)
filtered_sigma2 = gaussian_filter1d(lead_ii, sigma=sigma2)
filtered_sigma3 = gaussian_filter1d(lead_ii, sigma=sigma3)

# Plot results
plt.figure(figsize=(12, 10))

# Original ECG
plt.subplot(4,1,1)
plt.plot(time, lead_ii, color='blue')
plt.title("Original Lead II ECG")
plt.ylabel("Voltage (µV)")
plt.grid(True)

# Sigma = 1
plt.subplot(4,1,2)
plt.plot(time, filtered_sigma1, color='green')
plt.title("Gaussian Filter (σ = 1)")
plt.ylabel("Voltage (µV)")
plt.grid(True)

# Sigma = 3
plt.subplot(4,1,3)
plt.plot(time, filtered_sigma2, color='orange')
plt.title("Gaussian Filter (σ = 3)")
plt.ylabel("Voltage (µV)")
plt.grid(True)

# Sigma = 5
plt.subplot(4,1,4)
plt.plot(time, filtered_sigma3, color='red')
plt.title("Gaussian Filter (σ = 5)")
plt.xlabel("Time (s)")
plt.ylabel("Voltage (µV)")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
### Apply Butterworth Low Pass filter

from scipy.signal import butter, filtfilt


# Extract first 3 seconds of Lead II
duration = 3
num_samples = fs * duration

lead_ii = df["II"].values[:num_samples]

# Time axis
time = np.arange(num_samples) / fs

# -----------------------------------
# Butterworth Low-Pass Filter Function
# -----------------------------------
def butter_lowpass_filter(signal, cutoff, fs, order=4):
    nyquist = 0.5 * fs
    normal_cutoff = cutoff / nyquist
    b, a = butter(order, normal_cutoff, btype='low')
    filtered_signal = filtfilt(b, a, signal)
    return filtered_signal

# Apply filters
filtered_15 = butter_lowpass_filter(lead_ii, 15, fs)
filtered_20 = butter_lowpass_filter(lead_ii, 20, fs)
filtered_40 = butter_lowpass_filter(lead_ii, 40, fs)

# -----------------------------------
# Plot Results
# -----------------------------------
plt.figure(figsize=(12, 10))

# Original ECG
plt.subplot(4,1,1)
plt.plot(time, lead_ii, color='black')
plt.title("Original Lead II ECG")
plt.ylabel("Voltage (µV)")
plt.grid(True)

# 15 Hz
plt.subplot(4,1,2)
plt.plot(time, filtered_15, color='blue')
plt.title("Butterworth Low-Pass Filter (Cutoff = 15 Hz)")
plt.ylabel("Voltage (µV)")
plt.grid(True)

# 20 Hz
plt.subplot(4,1,3)
plt.plot(time, filtered_20, color='green')
plt.title("Butterworth Low-Pass Filter (Cutoff = 20 Hz)")
plt.ylabel("Voltage (µV)")
plt.grid(True)

# 40 Hz
plt.subplot(4,1,4)
plt.plot(time, filtered_40, color='red')
plt.title("Butterworth Low-Pass Filter (Cutoff = 40 Hz)")
plt.xlabel("Time (s)")
plt.ylabel("Voltage (µV)")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
### Apply Butterworth Band Pass filter

from scipy.signal import butter, filtfilt


# Extract first 3 seconds of Lead II
duration = 3
num_samples = fs * duration

lead_ii = df["II"].values[:num_samples]

# Time axis
time = np.arange(num_samples) / fs

# -----------------------------------------
# Butterworth Band-Pass Filter Function
# -----------------------------------------
def butter_bandpass_filter(signal, lowcut, highcut, fs, order=4):
    nyquist = 0.5 * fs
    low = lowcut / nyquist
    high = highcut / nyquist

    b, a = butter(order, [low, high], btype='band')
    filtered_signal = filtfilt(b, a, signal)

    return filtered_signal

# Apply Band-Pass Filter (0.5–40 Hz)
filtered_ecg = butter_bandpass_filter(
    lead_ii,
    lowcut=0.5,
    highcut=40,
    fs=fs,
    order=4
)

# -----------------------------------------
# Plot Original and Filtered ECG
# -----------------------------------------
plt.figure(figsize=(12,8))

# Original ECG
plt.subplot(2,1,1)
plt.plot(time, lead_ii, color='blue')
plt.title("Original Lead II ECG")
plt.xlabel("Time (s)")
plt.ylabel("Voltage (µV)")
plt.grid(True)

# Filtered ECG
plt.subplot(2,1,2)
plt.plot(time, filtered_ecg, color='red')
plt.title("Butterworth Band-Pass Filtered ECG (0.5–40 Hz)")
plt.xlabel("Time (s)")
plt.ylabel("Voltage (µV)")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
## Apply Notch filter

from scipy.signal import iirnotch, filtfilt

# Sampling frequency (Hz)
fs = 500

# Extract first 3 seconds of Lead II
duration = 3
num_samples = fs * duration

lead_ii = df["II"].values[:num_samples]

# Create time axis
time = np.arange(num_samples) / fs

# ------------------------------------
# Design Notch Filter
# ------------------------------------
notch_freq = 50      # Use 60 if your region uses 60 Hz power
Q = 30               # Quality factor

# Design filter
b, a = iirnotch(notch_freq, Q, fs)

# Apply zero-phase filtering
filtered_ecg = filtfilt(b, a, lead_ii)

# ------------------------------------
# Plot Original and Filtered ECG
# ------------------------------------
plt.figure(figsize=(12,8))

# Original ECG
plt.subplot(2,1,1)
plt.plot(time, lead_ii, color='blue')
plt.title("Original Lead II ECG")
plt.xlabel("Time (s)")
plt.ylabel("Voltage (µV)")
plt.grid(True)

# Filtered ECG
plt.subplot(2,1,2)
plt.plot(time, filtered_ecg, color='red')
plt.title(f"Lead II ECG After {notch_freq} Hz Notch Filter")
plt.xlabel("Time (s)")
plt.ylabel("Voltage (µV)")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
### Display all the ECG signal above using multiple subplots

from scipy.signal import butter, filtfilt, medfilt, savgol_filter, iirnotch
from scipy.ndimage import gaussian_filter1d

# -------------------------------
# Load ECG data
# -------------------------------
file_path = "MUSE_20180111_155859_36000.csv"
df = pd.read_csv(file_path)

# Sampling frequency
fs = 500

# Extract first 3 seconds of Lead II
duration = 3
num_samples = fs * duration

lead_ii = df["II"].values[:num_samples]
time = np.arange(num_samples) / fs

# -------------------------------------------------
# 1. Moving Average Filter (Window = 5)
# -------------------------------------------------
window_size = 5
moving_avg = np.convolve(
    lead_ii,
    np.ones(window_size) / window_size,
    mode='same'
)

# -------------------------------------------------
# 2. Median Filter (Kernel = 3)
# -------------------------------------------------
median_filtered = medfilt(lead_ii, kernel_size=3)

# -------------------------------------------------
# 3. Gaussian Filter (Sigma = 2)
# -------------------------------------------------
gaussian_filtered = gaussian_filter1d(lead_ii, sigma=2)

# -------------------------------------------------
# 4. Savitzky-Golay Filter
# -------------------------------------------------
savgol_filtered = savgol_filter(
    lead_ii,
    window_length=11,
    polyorder=2
)

# -------------------------------------------------
# 5. Butterworth Low-Pass Filter
# -------------------------------------------------
def butter_lowpass(signal, cutoff, fs, order=4):
    nyquist = 0.5 * fs
    b, a = butter(order, cutoff / nyquist, btype='low')
    return filtfilt(b, a, signal)

lowpass_filtered = butter_lowpass(
    lead_ii,
    cutoff=20,
    fs=fs
)

# -------------------------------------------------
# 6. Butterworth Band-Pass Filter
# -------------------------------------------------
def butter_bandpass(signal, lowcut, highcut, fs, order=4):
    nyquist = 0.5 * fs
    low = lowcut / nyquist
    high = highcut / nyquist

    b, a = butter(order, [low, high], btype='band')
    return filtfilt(b, a, signal)

bandpass_filtered = butter_bandpass(
    lead_ii,
    lowcut=0.5,
    highcut=40,
    fs=fs
)

# -------------------------------------------------
# 7. Notch Filter (50 Hz)
# -------------------------------------------------
notch_freq = 50
Q = 30

b, a = iirnotch(notch_freq, Q, fs)
notch_filtered = filtfilt(b, a, lead_ii)

# -------------------------------------------------
# Plot all signals
# -------------------------------------------------
fig, axes = plt.subplots(
    8,
    1,
    figsize=(14,16),
    sharex=True
)

# Original
axes[0].plot(time, lead_ii, color='black')
axes[0].set_title("Original ECG")
axes[0].set_ylabel("µV")
axes[0].grid(True)

# Moving Average
axes[1].plot(time, moving_avg, color='blue')
axes[1].set_title("Moving Average Filter (Window = 5)")
axes[1].set_ylabel("µV")
axes[1].grid(True)

# Median
axes[2].plot(time, median_filtered, color='green')
axes[2].set_title("Median Filter (Kernel = 3)")
axes[2].set_ylabel("µV")
axes[2].grid(True)

# Gaussian
axes[3].plot(time, gaussian_filtered, color='orange')
axes[3].set_title("Gaussian Filter (σ = 2)")
axes[3].set_ylabel("µV")
axes[3].grid(True)

# Savitzky-Golay
axes[4].plot(time, savgol_filtered, color='purple')
axes[4].set_title("Savitzky–Golay Filter")
axes[4].set_ylabel("µV")
axes[4].grid(True)

# Butterworth Low-pass
axes[5].plot(time, lowpass_filtered, color='red')
axes[5].set_title("Butterworth Low-Pass Filter (20 Hz)")
axes[5].set_ylabel("µV")
axes[5].grid(True)

# Butterworth Band-pass
axes[6].plot(time, bandpass_filtered, color='brown')
axes[6].set_title("Butterworth Band-Pass Filter (0.5–40 Hz)")
axes[6].set_ylabel("µV")
axes[6].grid(True)

# Notch Filter
axes[7].plot(time, notch_filtered, color='magenta')
axes[7].set_title("50 Hz Notch Filter")
axes[7].set_xlabel("Time (seconds)")
axes[7].set_ylabel("µV")
axes[7].grid(True)

plt.suptitle("Comparison of ECG Filtering Techniques", fontsize=16)

plt.tight_layout(rect=[0, 0, 1, 0.98])
plt.show()
```
